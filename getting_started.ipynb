### 2. Basic Computations and Plotting
Let's run a quick matrix calculation and plot a simple sine wave to check numpy and matplotlib visualization.

```python
# Quick NumPy test
matrix = np.random.rand(3, 3)
print("Random 3x3 Matrix:\n", matrix)
print("Eigenvalues:\n", np.linalg.eigvals(matrix))

# Quick Matplotlib/Seaborn test
sns.set_theme(style="darkgrid")
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(figsize=(8, 4))
plt.plot(x, y, label="$\\sin(x)$", color="#1f77b4", linewidth=2)
plt.title("Verification Plot: Sine Wave")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```
